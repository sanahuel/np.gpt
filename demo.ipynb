### Load Dataset and Simple Tokenizer

```python
from gpt.tokenizer import SimpleTokenizer
import numpy as np

dataset_path = './tinyshakespeare.txt'

with open(dataset_path, 'r', encoding='utf-8') as f:
    text = f.read()

tokenizer = SimpleTokenizer(text=text)
```

```python
tokens = tokenizer.encode("Lorem ipsum dolor sit...")
print(f"Encoded sequence: {tokens}")

string = tokenizer.decode(tokens)
print(f"Decoded sequence: {string}")
```

### Load Model

```python
from gpt.model import GPT

# Parameters
vocab_size = 65
d_model = 256
num_heads = 8
d_ff = 512
num_blocks = 4
batch_size = 1
sequence_length = 500
lr = 0.001

# Initialize model
model = GPT(vocab_size=vocab_size, 
            d_model=d_model,
            max_seq_len=sequence_length,
            num_heads=num_heads,
            d_ff=d_ff,
            num_blocks=num_blocks,
            lr=lr)
```

### Preprocess Dataset

```python
dataset = tokenizer.encode(text)
print(f"{len(dataset)} tokens in total")

train_tokes = dataset[:501]

train_data = []
for i in range(0, len(train_tokes) - sequence_length):
    x = train_tokes[i:i + sequence_length]
    y = train_tokes[i + 1:i + sequence_length + 1]
    train_data.append((x, y))
```

### Without Training

```python
string = 'First'
context = tokenizer.encode(string)
context = np.array(context, dtype=np.int32).reshape(1, -1)  # Add batch dimension

max_context_length = sequence_length - len(context[0]) + 1      # Remaining space for generated tokens
generated_tokens = model.generate(context, max_length=max_context_length, temperature=0.2)
generated_text = tokenizer.decode(generated_tokens.flatten())
print(generated_text)
```

### Training

```python
model.train(train_data, num_epochs=1000, batch_size=batch_size, lr=0.00001)
```

### Generate

```python
string = 'First'
context = tokenizer.encode(string)
context = np.array(context, dtype=np.int32).reshape(1, -1)  # Add batch dimension

max_context_length = sequence_length - len(context[0]) + 1      # Remaining space for generated tokens
generated_tokens = model.generate(context, max_length=max_context_length, temperature=0.5)
generated_text = tokenizer.decode(generated_tokens.flatten())
print(generated_text)
```
